# Validaciones de Calidad de Datos - Bronze

### Registro Mercantil (`workspace.bronze.bronze_registro_mercantil`)
1. Conteo total de registros
2. Volumen de nulos por columna
3. Duplicados exactos

8. Duplicados exactos

12. Consistencia de fechas (inicio ≤ fin)

```sql
-- 3. DUPLICADOS EXACTOS (filas 100% idénticas) - Registro Mercantil
SELECT COUNT(*) AS filas_duplicadas
FROM (
  SELECT *
  FROM workspace.bronze.bronze_registro_mercantil
  EXCEPT
  SELECT DISTINCT *
  FROM workspace.bronze.bronze_registro_mercantil
) d;
```

```sql
-- 4. FECHAS FUTURAS - Registro Mercantil
-- Las fechas en RM están como string YYYYMMDD. Se compara contra la fecha actual (2026-09-18 = 20260918)
SELECT 'fecha_matricula' AS columna,
       SUM(CASE WHEN fecha_matricula > '20260918' THEN 1 ELSE 0 END) AS fechas_futuras
FROM workspace.bronze.bronze_registro_mercantil
WHERE fecha_matricula IS NOT NULL AND TRIM(fecha_matricula) <> ''

UNION ALL
SELECT 'fecha_cancelacion',
       SUM(CASE WHEN fecha_cancelacion > '20260918' THEN 1 ELSE 0 END)
FROM workspace.bronze.bronze_registro_mercantil
WHERE fecha_cancelacion IS NOT NULL AND TRIM(fecha_cancelacion) <> ''

UNION ALL
SELECT 'fecha_vigencia',
       SUM(CASE WHEN fecha_vigencia > '20260918' THEN 1 ELSE 0 END)
FROM workspace.bronze.bronze_registro_mercantil
WHERE fecha_vigencia IS NOT NULL AND TRIM(fecha_vigencia) <> '';
```

```sql
-- 5. CONSISTENCIA TEMPORAL - Registro Mercantil
-- fecha_matricula no debería ser posterior a fecha_cancelacion (si la matrícula está cancelada)
SELECT 'fecha_matricula > fecha_cancelacion' AS inconsistencia,
       COUNT(*) AS num_registros
FROM workspace.bronze.bronze_registro_mercantil
WHERE fecha_matricula IS NOT NULL AND TRIM(fecha_matricula) <> ''
  AND fecha_cancelacion IS NOT NULL AND TRIM(fecha_cancelacion) <> ''
  AND fecha_matricula > fecha_cancelacion;
```

##Validaciones para SECOP

```sql
-- 6. CONTEO TOTAL DE REGISTROS - SECOP Contratos
SELECT COUNT(*) AS total_registros
FROM workspace.bronze.bronze_secop_contratos;
```

```sql
%python
# 7. VOLUMEN DE NULOS O VACIOS POR COLUMNA - SECOP Contratos
# SECOP tiene 87+ columnas, se usa PySpark para iterar eficientemente
df = spark.table("workspace.bronze.bronze_secop_contratos")
total = df.count()

from pyspark.sql.functions import col, trim, when, sum as spark_sum
from pyspark.sql.types import StringType

null_counts = []
for c in df.columns:
    if isinstance(df.schema[c].dataType, StringType):
        nulls = df.select(spark_sum(when(col(c).isNull() | (trim(col(c)) == ""), 1).otherwise(0)).alias("n")).collect()[0]["n"]
    else:
        nulls = df.select(spark_sum(when(col(c).isNull(), 1).otherwise(0)).alias("n")).collect()[0]["n"]
    null_counts.append((c, nulls, total))

null_counts.sort(key=lambda x: x[1], reverse=True)

import pandas as pd
result = pd.DataFrame(null_counts, columns=["columna", "nulos", "total"])
result["pct_nulos"] = (result["nulos"] / result["total"] * 100).round(2)
display(result)
```

```sql
%python
# 8. DUPLICADOS EXACTOS (filas 100% idénticas) - SECOP Contratos
# urlproceso es tipo MAP, no soporta EXCEPT en SQL. Se usa PySpark dropDuplicates
from pyspark.sql.functions import col
df = spark.table("workspace.bronze.bronze_secop_contratos")
# Cast MAP column to STRING for duplicate detection
df = df.withColumn("urlproceso", col("urlproceso").cast("string"))
total = df.count()
distinct = df.dropDuplicates().count()
duplicates = total - distinct

print(f"Total filas: {total}")
print(f"Filas distintas: {distinct}")
print(f"Filas duplicadas: {duplicates}")
```

```sql
-- 9. FECHAS FUTURAS - SECOP Contratos
-- Las fechas en SECOP están como string en formato ISO (YYYY-MM-DD...)
-- Se compara contra la fecha actual 2026-09-18
SELECT 'fecha_de_firma' AS columna,
       SUM(CASE WHEN fecha_de_firma > '2026-09-18' THEN 1 ELSE 0 END) AS fechas_futuras
FROM workspace.bronze.bronze_secop_contratos
WHERE fecha_de_firma IS NOT NULL AND TRIM(fecha_de_firma) <> ''

UNION ALL
SELECT 'fecha_de_inicio_del_contrato',
       SUM(CASE WHEN fecha_de_inicio_del_contrato > '2026-09-18' THEN 1 ELSE 0 END)
FROM workspace.bronze.bronze_secop_contratos
WHERE fecha_de_inicio_del_contrato IS NOT NULL AND TRIM(fecha_de_inicio_del_contrato) <> ''

UNION ALL
SELECT 'fecha_de_fin_del_contrato',
       SUM(CASE WHEN fecha_de_fin_del_contrato > '2026-09-18' THEN 1 ELSE 0 END)
FROM workspace.bronze.bronze_secop_contratos
WHERE fecha_de_fin_del_contrato IS NOT NULL AND TRIM(fecha_de_fin_del_contrato) <> '';
```

```sql
-- 10. VALORES NEGATIVOS O CERO DONDE NO APLICA - SECOP Contratos
-- valor_del_contrato, valor_pagado, valor_facturado están como string; se usa TRY_CAST
SELECT 'valor_del_contrato' AS columna,
       SUM(CASE WHEN TRY_CAST(valor_del_contrato AS DECIMAL(38,2)) < 0 THEN 1 ELSE 0 END) AS negativos,
       SUM(CASE WHEN TRY_CAST(valor_del_contrato AS DECIMAL(38,2)) = 0 THEN 1 ELSE 0 END) AS en_cero
FROM workspace.bronze.bronze_secop_contratos
WHERE valor_del_contrato IS NOT NULL AND TRIM(valor_del_contrato) <> ''

UNION ALL
SELECT 'valor_pagado',
       SUM(CASE WHEN TRY_CAST(valor_pagado AS DECIMAL(38,2)) < 0 THEN 1 ELSE 0 END),
       SUM(CASE WHEN TRY_CAST(valor_pagado AS DECIMAL(38,2)) = 0 THEN 1 ELSE 0 END)
FROM workspace.bronze.bronze_secop_contratos
WHERE valor_pagado IS NOT NULL AND TRIM(valor_pagado) <> ''

UNION ALL
SELECT 'valor_facturado',
       SUM(CASE WHEN TRY_CAST(valor_facturado AS DECIMAL(38,2)) < 0 THEN 1 ELSE 0 END),
       SUM(CASE WHEN TRY_CAST(valor_facturado AS DECIMAL(38,2)) = 0 THEN 1 ELSE 0 END)
FROM workspace.bronze.bronze_secop_contratos
WHERE valor_facturado IS NOT NULL AND TRIM(valor_facturado) <> '';
```

```sql
-- 11. OUTLIERS EN valor_del_contrato - SECOP Contratos
-- Percentiles para detectar valores atipicos extremos
SELECT
  MIN(TRY_CAST(valor_del_contrato AS DECIMAL(38,2))) AS valor_min,
  PERCENTILE(TRY_CAST(valor_del_contrato AS DECIMAL(38,2)), 0.25) AS p25,
  PERCENTILE(TRY_CAST(valor_del_contrato AS DECIMAL(38,2)), 0.50) AS mediana,
  PERCENTILE(TRY_CAST(valor_del_contrato AS DECIMAL(38,2)), 0.75) AS p75,
  PERCENTILE(TRY_CAST(valor_del_contrato AS DECIMAL(38,2)), 0.95) AS p95,
  PERCENTILE(TRY_CAST(valor_del_contrato AS DECIMAL(38,2)), 0.99) AS p99,
  MAX(TRY_CAST(valor_del_contrato AS DECIMAL(38,2))) AS valor_max,
  COUNT(*) AS total_contratos_con_valor
FROM workspace.bronze.bronze_secop_contratos
WHERE valor_del_contrato IS NOT NULL AND REGEXP_LIKE(TRIM(valor_del_contrato), '^[0-9]+(\\.[0-9]+)?$');
```

```sql
-- 12. CONSISTENCIA DE FECHAS (inicio <= fin) - SECOP Contratos
-- fecha_de_inicio_del_contrato no debería ser posterior a fecha_de_fin_del_contrato
SELECT 'fecha_inicio > fecha_fin' AS inconsistencia,
       COUNT(*) AS num_registros
FROM workspace.bronze.bronze_secop_contratos
WHERE fecha_de_inicio_del_contrato IS NOT NULL AND TRIM(fecha_de_inicio_del_contrato) <> ''
  AND fecha_de_fin_del_contrato IS NOT NULL AND TRIM(fecha_de_fin_del_contrato) <> ''
  AND fecha_de_inicio_del_contrato > fecha_de_fin_del_contrato;
```
